```python
### Given Data

image = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

kernel = [
    [1, 0],
    [0, 1]
]
```

```python
def convolve(image, kernel):
    # Get the size of the image and kernel
    img_size = len(image)
    k_size = len(kernel)

    # Calculate the size of the output
    out_size = img_size - k_size + 1

    # Create an empty list to store the output
    output = []

    # Loop through each possible position of the kernel
    for i in range(out_size):
        row = []

        for j in range(out_size):
            total = 0

            # Multiply overlapping values and add them together
            for x in range(k_size):
                for y in range(k_size):
                    total += image[i + x][j + y] * kernel[x][y]

            # Store the result for this position
            row.append(total)

        # Add the completed row to the output
        output.append(row)

    return output
```

```python
result = convolve(image, kernel)

for row in result:
    print(row)
```
